## Exlporatory Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("../dataset/Algerian_forest_fires_dataset_CLEANED.csv")
```

```python
## drop day, month and year
df.drop(["day","month","year"], axis=1)
```

```python
df.head()
```

```python
## Encoding of the categories in classes
df["Classes"] = np.where(df["Classes"].str.contains("not fire"), 0, 1)
```

```python
df.tail()
```

```python
df["Classes"].value_counts()
```

```python
# Plotting density plot for all features
plt.style.use("fivethirtyeight")
df.hist(bins = 40, figsize = (30,15))
plt.show()
```

```python
# Percentage for Pie Chart

plt.pie(df["Classes"].value_counts(), labels = ["Not Fire", "Fire"], autopct = "%1.1f%%", shadow = True, startangle = 90)
plt.title("Percentage of Fire and Not Fire")
plt.show()
```

```python
## Correlation
df.corr()
```

```python
sns.heatmap(df.corr())
```

```python
# Box Plots
sns.boxplot(df['FWI'], color = 'green')
```

```python
df.head()
```

```python
## Monthly Fire Analysis
dftemp = df.loc[df['Region']==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue="Classes",data = df)
plt.title("Fire Analysis of Bejaia Regions", weight = 'bold')
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight = 'bold')
plt.show()
```

```python
## Monthly Fire Analysis
dftemp = df.loc[df['Region']==0]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month',hue="Classes",data = df)
plt.title("Fire Analysis of Sidi-Bel Regions", weight = 'bold')
plt.ylabel('Number of Fires',weight='bold')
plt.xlabel('Months',weight = 'bold')
plt.show()
```

It is observed that August and September had the most number of forest fires for both regions. And from the above plots, we can understand few things

Most of the fires happened in August and very high Fres happened in only 3 months - June July and August.

Less Fires on September.
